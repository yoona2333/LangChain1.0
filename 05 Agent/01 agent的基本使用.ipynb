```python
# 旧版社区包，类名 TavilySearchResults
from langchain_community.tools.tavily_search import TavilySearchResults

from mimetypes import init
import os

from langchain_tavily import TavilySearch
from rich import print as rprint
# 后面某处写了 rprint = something 把rprint覆盖成模块了
from langchain.tools import tool

from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model

load_dotenv(override=True)
os.getenv("CLOSEAI_API_KEY")
CLOSEAI_API_KEY = os.getenv("CLOSEAI_API_KEY")
CLOSEAI_BASE_URL = os.getenv("CLOSEAI_BASE_URL")


@tool
def get_weather(city: str) -> str:
    """
    获取城市的天气
    """
    return f"{city}的天气是晴朗"


# web搜索工具 使用内置的TavilySearch
web_search=TavilySearch(
    api_key=os.getenv("tavily_api_key"),
    max_results=2
)

model = init_chat_model(
    model="deepseek-flash",
    api_key=CLOSEAI_API_KEY,
    base_url=CLOSEAI_BASE_URL,
)
agent = create_agent(
    model=model,
    tools=[get_weather,web_search],

)
# print(type(agent))



response = agent.invoke({
    "messages": [
        {"role": "user", "content":"帮我查询25年诺贝尔物理学奖的获奖者"}
    ]
})

rprint(response["messages"][-1].content)


```
